```python
import numpy as np

def lfsr(polymer, state):
    """Simulates a Linear Feedback Shift Register."""
    state = list(state)
    while True:
        yield state[-1]
        # Calculate feedback bit (XOR of taps)
        feedback = 0
        for i in range(len(polymer)):
            if polymer[i] == 1:
                feedback ^= state[i]
        # Shift and insert feedback
        state = [feedback] + state[:-1]

def generate_gold_codes(num_users=4, length=4):
    # Preferred pair of polynomials for n=5 (length 31)
    # p1: x^5 + x^2 + 1 -> [1, 0, 0, 1, 0]
    # p2: x^5 + x^4 + x^3 + x^2 + 1 -> [1, 1, 1, 1, 0]
    poly1 = [1, 0, 0, 1, 0]
    poly2 = [1, 1, 1, 1, 0]
    
    # Initial states (must be non-zero)
    reg1 = lfsr(poly1, [1] * length)
    reg2 = lfsr(poly2, [1] * length)
    
    # Generate the two base m-sequences
    m_seq1 = [next(reg1) for _ in range(2**length - 1)]
    m_seq2 = [next(reg2) for _ in range(2**length - 1)]
    
    gold_codes = []
    
    # Generate codes by XORing m_seq1 with shifted versions of m_seq2
    for shift in range(num_users):
        shifted_m2 = m_seq2[shift:] + m_seq2[:shift]
        gold_code = [(a ^ b) for a, b in zip(m_seq1, shifted_m2)]
        # Convert 0/1 to -1/1 for BPSK modulation
        bpsk_code = [1 if x == 0 else 0 for x in gold_code]
        gold_codes.append(bpsk_code)
        
    return gold_codes

# Execution
users = 2
codes = generate_gold_codes(num_users=users)
for i, code in enumerate(codes):
    print(f"User {i+1}: {code}")
```
