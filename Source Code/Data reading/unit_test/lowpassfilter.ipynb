```python
import numpy as np
import os
import matplotlib.pyplot as plt
#os.chdir(os.path.dirname(os.path.abspath(__file__)))
filename = "multi_BPSK.npy"
rx_signal = np.load(filename).astype(float)[2000:3000]
plt.figure(figsize=(15, 4))
plt.plot(rx_signal)
out=np.zeros(len(rx_signal))
out[0]=1884
for i in range(len(rx_signal)-1):
    out[i+1]=out[i]*15/16+rx_signal[i+1]/16
plt.plot(out)
```
